## Prace przy modelu uczenia maszynowego

```python
# importy danych
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
import os
from dotenv import load_dotenv
from pathlib import Path

# pobranie zmiennych env
load_dotenv()
test_size = float(os.getenv("TEST_SIZE"))
random_seed = int(os.getenv("RANDOM_SEED"))
output_dir = Path(os.getenv("REPORT_PATH"))
output_dir.mkdir(parents=True, exist_ok=True)
```

```python
# zapisanie danych w dataframe
data = load_iris(as_frame=True)
iris_features = data.data
iris_target = data.target

# wyświetlenie podstawowych informacji o zbiorze
display(iris_features.head())
print(f"\nRozmiar wybranego zbioru: {iris_features.shape[0]}x{iris_features.shape[1]}\n")
display(iris_features.describe())

print(f"Klasy w zbiorze: {iris_target.unique()}")
```

```python
# podział danych na zbiory treningowy i testowy
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(iris_features, iris_target, test_size=test_size, random_state=random_seed)
```

```python
# wykorzystanie drzewa decyzyjnego
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, accuracy_score
clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)

y_preds = clf.predict(X_test)
accuracy = accuracy_score(y_test, y_preds)
print("Jakość predykcji: ", accuracy)
print("\nWypisanie raportu:")
print(classification_report(y_test, y_preds))
```

```python
# zapisanie classification_report do pliku .txt
report = classification_report(y_test, y_preds)

(output_dir / "report_card.txt").write_text(
    report,
    encoding="utf-8"
)
```
